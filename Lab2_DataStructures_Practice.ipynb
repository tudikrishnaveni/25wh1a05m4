### 7. Dictionary Comprehension
Using the `gradebook` from Exercise 6, build a simple `{subject: marks}` dictionary in **one line** using a dictionary comprehension.

```python
# TODO: 7. Dictionary comprehension solution
subject={
    "python":28,
    "ds":29,
    "maths":30,
    "aep":29,
    "dbms":29
}

gradebook = {
    subject: {"marks": marks}
   for subject, marks in subject.items()
}

print(gradebook)
```

## Combined Task — Semester Report (submit this)

Bring everything together into one program that:
1. Defines a nested `gradebook` **dictionary** (at least 6 subjects, with `marks` and `credits`).
2. Defines a `lab_submissions` **list** with at least 10 entries (with repeats).
3. Computes the **unique subjects submitted** using a **set**.
4. Computes the **high-scoring subjects (marks >= 85)** using a **set comprehension**.
5. Computes the **best subject** as a **tuple** `(subject_name, marks)` — the highest-scoring subject overall.
6. Prints all of the above in a clearly labelled semester report.


```python
# TODO: Combined task solution

gradebook={
    "python":{
        "marks":58,
        "credits":2,

    },
      "DS":{

        "marks":99,
        "credits":2,

    },
    "maths":{

        "marks":90,
        "credits":2,

    },
    "aep":{

        "marks":89,
        "credits":2,
         },
    "dbms":{

        "marks":89,
        "credits":2,

    }
}
lab_submissions=["C","JAVA","PYTHON","DBMS","OS","C++","PYTHON","DBMS","OS","C++"]
uniquesub=set(lab_submissions)
high_scoring_subjects = {
    sub
    for sub, details in gradebook.items()
    if details["marks"] >= 85
}
best_subject = None
highest_marks = -1

for sub, details in gradebook.items():
    if details["marks"] > highest_marks:
        highest_marks = details["marks"]
        best_subject = (sub, details["marks"])

print("---------semester report----------")
print(f"unique subject:{uniquesub}")
print(f"high scoring subjects:{high_scoring_subjects}")
print(f"best subject:{best_subject}")
```

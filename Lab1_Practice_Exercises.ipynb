````python
# TODO: B10 solution
def bill_and_collect_cash(cart, menu, cash_paid, is_parcel=False):
    try:
        bill = calculate_bill(cart, menu, is_parcel)
        if cash_paid < bill["total"]:
            raise ValueError(f"Insufficient cash: paid Rs.{cash_paid}, bill is Rs.{bill['total']}")
        change = round(cash_paid - bill["total"], 2)
        print(f"Bill: Rs.{bill['total']} | Change: Rs.{change}")
        return bill
    except KeyError as e:
        print(f"Item not on menu: {e}")
    except ValueError as e:
        print(f"Transaction failed: {e}")

bill_and_collect_cash(["samosa", "Dosa"], menu, cash_paid=50)
bill_and_collect_cash(["samosa", "tea"], menu, cash_paid=60)
bill_and_collect_cash(["water", "tea"], menu, cash_paid=35)
````

## Part C — End-of-Lab Combined Task (submit this)

Bring B1–B10 together into one working canteen billing program that:
1. Defines a **menu dictionary** with at least 6 items.
2. Lets you build a **cart (list)** for a customer.
3. Uses your **`calculate_bill()` function** (from B8) to bill the cart, including discount and packing charge logic.
4. **Saves** the receipt to `my_canteen_sales_log.txt` (from B9).
5. Wraps the whole transaction in **try/except** (from B10) so an invalid item or insufficient cash doesn't crash the program.
6. Bills **at least 3 different customers/carts** in a row, appending each to the same log file, then prints the full day's log at the end.


````python
# TODO: Part C solution
# Canteen Billing Program - B1 to B10 Combined

menu = {
    "idli": 30,
    "dosa": 50,
    "vada": 25,
    "poori": 45,
    "rice": 60,
    "tea": 15,
    "coffee": 20
}

LOG_FILE = "my_canteen_sales_log.txt"


def calculate_bill(cart):
    """Calculate subtotal, discount, packing charge, and final total."""
    subtotal = sum(menu[item] for item in cart)

    # 10% discount for bills of Rs. 200 or more
    discount = subtotal * 0.10 if subtotal >= 200 else 0

    # Packing charge of Rs. 20 for bills below Rs. 500
    packing_charge = 20 if subtotal < 500 else 0

    total = subtotal - discount + packing_charge

    return subtotal, discount, packing_charge, total


def save_receipt(customer, cart, subtotal, discount, packing_charge, total, cash):
    """Append one customer's receipt to the sales log."""
    balance = cash - total

    with open(LOG_FILE, "a") as file:
        file.write("\n" + "=" * 40 + "\n")
        file.write(f"Customer: {customer}\n")
        file.write("Items:\n")

        for item in cart:
            file.write(f"  {item.title()} - Rs.{menu[item]:.2f}\n")

        file.write(f"Subtotal:       Rs.{subtotal:.2f}\n")
        file.write(f"Discount:       Rs.{discount:.2f}\n")
        file.write(f"Packing Charge: Rs.{packing_charge:.2f}\n")
        file.write(f"Total:          Rs.{total:.2f}\n")
        file.write(f"Cash Paid:      Rs.{cash:.2f}\n")
        file.write(f"Change:         Rs.{balance:.2f}\n")
        file.write("=" * 40 + "\n")


def bill_customer(customer, cart, cash):
    """Process one complete customer transaction."""
    try:
        # Validate all items before calculating the bill
        for item in cart:
            if item not in menu:
                raise ValueError(f"Invalid menu item: {item}")

        subtotal, discount, packing_charge, total = calculate_bill(cart)

        if cash < total:
            raise ValueError(
                f"Insufficient cash. Required Rs.{total:.2f}, "
                f"but received Rs.{cash:.2f}"
            )

        save_receipt(
            customer,
            cart,
            subtotal,
            discount,
            packing_charge,
            total,
            cash
        )

        print(f"\nCustomer: {customer}")
        print(f"Items: {', '.join(cart)}")
        print(f"Total: Rs.{total:.2f}")
        print(f"Cash: Rs.{cash:.2f}")
        print(f"Change: Rs.{cash - total:.2f}")
        print("Transaction successful.")

    except (ValueError, TypeError) as error:
        print(f"\nTransaction failed for {customer}: {error}")


# Start with a fresh log for today's transactions
open(LOG_FILE, "w").close()

# ---------------------------------------------------------
# Bill at least 3 different customers/carts in a row
# ---------------------------------------------------------

customer1_cart = ["idli", "dosa", "tea"]
customer1_cash = 200

customer2_cart = ["rice", "vada", "coffee", "dosa"]
customer2_cash = 300

customer3_cart = ["poori", "idli", "vada", "tea"]
customer3_cash = 200

bill_customer("Customer 1", customer1_cart, customer1_cash)
bill_customer("Customer 2", customer2_cart, customer2_cash)
bill_customer("Customer 3", customer3_cart, customer3_cash)

# Example of error handling:
# This will not crash the program because "pizza" is not on the menu.
bill_customer("Invalid Customer", ["dosa", "pizza"], 200)

# This will also be caught because the cash is insufficient.
bill_customer("Low Cash Customer", ["rice", "dosa"], 50)


# ---------------------------------------------------------
# Print the complete day's sales log
# ---------------------------------------------------------

print("\n\n========== FULL DAY'S SALES LOG ==========")

try:
    with open(LOG_FILE, "r") as file:
        print(file.read())
except FileNotFoundError:
    print("Sales log file not found.")

````

### Scenario
It's lunch rush. Multiple customers place orders in quick succession. Some orders will have mistakes (an item not on the menu, or not enough cash). Your program must process the whole rush, skip and log the bad orders, and still produce an accurate summary at the end.

### Setup
```python
menu = {

orders = [
```

### Requirements

Write `process_order(order, menu)` and a driver loop that, across all orders in `orders`:

1. **Peak-hour surcharge** — if `hour` is 12 or 13 (lunch rush), add an 8% surcharge on the subtotal *before* GST. Outside those hours, no surcharge.
2. **Two competing discounts, pick the better one** — a membership discount (`regular`: 0%, `silver`: 5%, `gold`: 12%) *and* the loyalty discount from earlier (subtotal > 200 → 10%, > 100 → 5%, else 0%). Apply whichever discount is **larger**, never both stacked.
3. **GST (5%)** applies after the discount, and a **Rs.5 packing charge** if `is_parcel` is True — same as before.
4. **Bad orders must not stop the run.** If a cart item isn't in `menu`, catch it (`KeyError`), log the order to `failed_transactions.txt` with the reason, and move on to the next order. Same for insufficient cash (`ValueError`) — don't let the loop die.
5. **Successful orders** get appended to `rush_hour_log.txt` as `"<customer>: Rs.<total>"`.
6. **End-of-rush summary**, printed (not just written to file):
   - Total revenue collected across all successful orders.
   - Revenue broken down **by category** (Snacks / Beverages / Meals) — build this without importing anything, using a plain dictionary you update as you go.
   - The single **best-selling item** by total quantity sold across all orders (again, no imports — track counts in a dictionary).
   - How many orders **failed**, and why.

Try to get through all 5 sample orders (including the two broken ones) without your program stopping partway.


````python
# TODO: Stretch challenge solution
menu = {
    "Samosa": {"price": 15.0, "category": "Snacks"},
    "Vada": {"price": 12.0, "category": "Snacks"},
    "Tea": {"price": 10.0, "category": "Beverages"},
    "Cold Coffee": {"price": 30.0, "category": "Beverages"},
    "Dosa": {"price": 40.0, "category": "Meals"},
    "Idli": {"price": 20.0, "category": "Meals"}
}
orders = [
    {"customer": "Ravi",   "cart": ["Samosa", "Tea"],       "tier": "regular", "hour": 13, "is_parcel": False, "cash_paid": 30},
    {"customer": "Ananya", "cart": ["Dosa", "Cold Coffee"], "tier": "gold",    "hour": 12, "is_parcel": True,  "cash_paid": 100},
    {"customer": "Kiran",  "cart": ["Vada", "Idli"],        "tier": "silver",  "hour": 14, "is_parcel": False, "cash_paid": 45},
    {"customer": "Meena",  "cart": ["Pizza"],               "tier": "regular", "hour": 13, "is_parcel": False, "cash_paid": 100},
    {"customer": "Arjun",  "cart": ["Dosa", "Dosa", "Tea"], "tier": "gold",    "hour": 12, "is_parcel": False, "cash_paid": 95},
    {"customer": "Divya",  "cart": ["Tea"],                 "tier": "regular", "hour": 15, "is_parcel": False, "cash_paid": 5},
]
def process_order(order, menu):

    subtotal = 0
    for item in order["cart"]:
        subtotal += menu[item]["price"]
    if order["hour"] in [12, 13]:
        subtotal += subtotal * 0.08


    membership = {
        "regular": 0,
        "silver": 5,
        "gold": 12
    }

    member_discount = membership[order["tier"]]


    if subtotal > 200:
        loyalty_discount = 10
    elif subtotal > 100:
        loyalty_discount = 5
    else:
        loyalty_discount = 0


    discount = max(member_discount, loyalty_discount)

    subtotal -= subtotal * discount / 100

    total = subtotal + subtotal * 0.05


    if order["is_parcel"]:
        total += 5

    total = round(total, 2)


    if order["cash_paid"] < total:
        raise ValueError("Insufficient cash")

    return total

revenue = 0

category_revenue = {
    "Snacks": 0,
    "Beverages": 0,
    "Meals": 0
}

item_count = {}

failed_orders = 0
failure_reason = {}


open("rush_hour_log.txt", "w").close()
open("failed_transactions.txt", "w").close()

for order in orders:

    try:
        total = process_order(order, menu)

        revenue += total
        for item in order["cart"]:

            category = menu[item]["category"]
            category_revenue[category] += menu[item]["price"]


            if item in item_count:
                item_count[item] += 1
            else:
                item_count[item] = 1

        with open("rush_hour_log.txt", "a") as file:
            file.write(f"{order['customer']}: Rs.{total}\n")

        print(order["customer"], "Order Successful")

    except KeyError as e:

        failed_orders += 1
        reason = f"Item not found: {e}"

        failure_reason[reason] = failure_reason.get(reason, 0) + 1

        with open("failed_transactions.txt", "a") as file:
            file.write(f"{order['customer']} : {reason}\n")

        print(order["customer"], "-", reason)

    except ValueError as e:

        failed_orders += 1
        reason = str(e)

        failure_reason[reason] = failure_reason.get(reason, 0) + 1

        with open("failed_transactions.txt", "a") as file:
            file.write(f"{order['customer']} : {reason}\n")

        print(order["customer"], "-", reason)



print("\n========== END OF RUSH SUMMARY ==========")

print("Total Revenue Collected : Rs.", round(revenue, 2))

print("\nRevenue By Category")

for category in category_revenue:
    print(category, ":", round(category_revenue[category], 2))

best_item = None
best_quantity = 0

for item in item_count:
    if item_count[item] > best_quantity:
        best_quantity = item_count[item]
        best_item = item

print("\nBest Selling Item :", best_item)
print("Quantity Sold :", best_quantity)

print("\nFailed Orders :", failed_orders)

print("Failure Reasons")

for reason in failure_reason:
    print(reason, ":", failure_reason[reason])
````
